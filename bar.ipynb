```python
import matplotlib.pyplot as plt
```

```python

movie=["dark","locker","speech","artists","Argo"]
years=[2003,2004,2005,2006,2007]
revenue=[34,56,78,23,39]

plt.bar(years,revenue,color="red")
plt.xlabel("Years")
plt.ylabel("Revenue in $M")
plt.title("Revenue in each year.")
# must take 4 args and ha stands for revenue
# +5  states the margin
# if it crossing the margin then limit can be uses
plt.ylim(0,max(revenue)+100)
for i in range(len(years)):
    plt.text(years[i],revenue[i]+5,str(revenue[i]),ha="center")

```
