```python
import matplotlib.pyplot as plt
```

```python


expenses = ["salaries", "rent", "marketing", "r&o", "extra"]
amount = [500, 150, 120, 100, 50]
#  autopct tell the percentage
plt.pie(
    amount,
    labels=expenses,
    autopct="%1.1f%%",     # shows percentage with 1 decimal
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1,
        "linestyle": "--"
    }
)

plt.title("Company Expenses")
plt.show()
```

```python
# explode->highlight a slice by pulling it out


expenses = ["salaries", "rent", "marketing", "r&o", "extra"]
amount = [500, 150, 120, 100, 50]
explode=[0,0,0.2,0,0]
#  autopct tell the percentage
plt.pie(
    amount,
    labels=expenses,
    autopct="%1.1f%%",     # shows percentage with 1 decimal
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1,
        "linestyle": "--"
    },
    explode=explode
)

plt.title("Company Expenses")
plt.show()
```

```python
# startangle for rotating 
# 180 ->opp
# 360 same


expenses = ["salaries", "rent", "marketing", "r&o", "extra"]
amount = [500, 150, 120, 100, 50]
explode=[0,0,0.2,0,0]
#  autopct tell the percentage
plt.pie(
    amount,
    labels=expenses,
    autopct="%1.1f%%",     # shows percentage with 1 decimal
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1,
        "linestyle": "--"
    },
    explode=explode,
    startangle=90,
    shadow=True
    
)

plt.title("Company Expenses")
plt.show()
```
