```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Aman", "Riya", "Kunal", "Neha", "Arjun", "Pooja"],
    "Age": [22, np.nan, 24, 28, np.nan, 23],
    "Experience": [1, 3, np.nan, 5, 4, np.nan],
    "Salary": [30000, 50000, 42000, np.nan, 65000, 32000],
    "Department": ["IT", "HR", "IT", np.nan, "IT", "HR"],
    "City": ["Delhi", "Mumbai", np.nan, "Bangalore", "Bangalore", np.nan]
}

df = pd.DataFrame(data)
print(df)
```

```python
df["City"].unique()
```

```python
df["new sal"]= df["Salary"]*20
df
```

```python
df["Salary"].isnull()
```

```python
df.dropna()
```

```python
df["Age"].fillna(df["Age"].mean(),inplace=True)
df
```

```python
df["Salary"].fillna(df["Salary"].median(),inplace=True)
df
```

```python
df["Department"].fillna(df["Department"].mode()[0],inplace=True)
df
```

```python
df.drop_duplicates()
df
```

```python
df["City"].fillna(df["City"].mode()[0],inplace=True)
df
```

```python
# lambda


df["Salary"]=df["Salary"].apply(lambda x :x/10 if x>50000 else x)
df
```
