```python
import pandas as pd
import numpy as np
```

```python
a=np.arange(0,20).reshape(5,4)
```

```python
# creating data frame
# first is data,second val is index names for rows then columns
df=pd.DataFrame(a,index=["row1","row2","row3","row4","row5"],columns=["col1","col2","col3","col4"])
df
```

```python
# using row index name using loc
df.loc[['row3','row4']]
```

```python
# accesng both using iloc ->index locaatio   
# row,index form

# index starts from 0 in both row and columns
df.iloc[2:4,0:2]
```

```python
# if only start index given then from that index till end

df.iloc[2:,1:]
```

```python
df.iloc[:,0]
```

```python
df.iloc[:,3]
```

```python
# combining in one
# : ->means all rows

df.iloc[:,[0,3]]
```
