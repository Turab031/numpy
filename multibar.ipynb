```python
import matplotlib.pyplot as plt
```

```python
oscar_movie=["dark","locker","speech","artists","Argo"]
years=[2003,2004,2005,2006,2007]
oscar_revenue=[34,56,78,23,39]
nonoscar_movie=["leech","pine","avatar","apple","termite"]
nonoscar_revenue=[23,50,73,20,30]

plt.title("oscar vs non-oscar revenue")


plt.xlabel("year")
plt.ylabel("revenue in $M")

plt.bar(years,oscar_revenue)
plt.bar(years,nonoscar_revenue)
```

```python
# overlapping as observed
# by default value is 0.8

import numpy as np 
x= np.arange(len(years))
oscar_movie=["dark","locker","speech","artists","Argo"]
years=[2003,2004,2005,2006,2007]
oscar_revenue=[34,56,78,23,39]
nonoscar_movie=["leech","pine","avatar","apple","termite"]
nonoscar_revenue=[23,50,73,20,30]

plt.title("oscar vs non-oscar revenue")


plt.xlabel("year")
plt.ylabel("revenue in $M")
width = 0.35
# shifting to left
plt.bar(x+width/2,oscar_revenue,width,label ="oscar movies")
# shifting to right
plt.bar(x-width/2,nonoscar_revenue,width,label="non-oscar")
plt.legend()
```
