```python
import matplotlib.pyplot as plt

people = ["a","b","c","d","e","f","g","h","i","J"]
age = [22,25,38,40,4,50,55,65,70,75]   # now 10 values
bp = [110,115,120,122,125,130,135,123,150,155]
colors=["red","purple","yellow","green","magenta","aqua","red","purple","yellow","green"]
colors= ["green" if x<135 else "red" for x in bp]
plt.scatter(age, bp,color=colors)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("BP")

# data annotation
for i in range(len(people)):
    plt.annotate(people[i],xy=(age[i],bp[i]),xytext=(age[i],bp[i]))
plt.grid()
plt.show()

```
