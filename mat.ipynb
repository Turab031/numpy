```python
oscar_movie=["dark","locker","speech","artists","Argo"]
years=[2003,2004,2008,2009,2010]
oscar_revenue=[34,56,78,23,39]
nonoscar_movie=["leech","pine","avatar","apple","termite"]
nonoscar_revenue=[23,50,73,20,30]
plt.plot(oscar_movie,oscar_revenue,label="oscar movie")
plt.plot(nonoscar_movie,nonoscar_revenue,label="non oscar")

plt.xlabel("year")
plt.ylabel("revenue in $M")
# legent gives line label/color
plt.legend()

plt.title("oscar vs non-oscar")

```
